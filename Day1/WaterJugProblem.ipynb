**Water Jug Problem**
>You are given two water Jugs:
 - Jug A = 4 litres
 - Jug B = 3 liters

Intially both are empty

The objective is to obtain exactly 2 liters of water jug A.

*We can't add water with measurement we have to use two jugs that can hold 4 and 3 litres*

```python

capacity_A = 4
capacity_B = 3
```

```python
def get_moves(state):
    a, b = state
    moves = []
    # Fill A
    moves.append((capacity_A, b))
    # Fill B
    moves.append((a, capacity_B))
    # Empty A
    moves.append((0, b))
    # Empty B
    moves.append((a, 0))
    # Pour A -> B
    transfer = min(a, capacity_B - b)
    moves.append((a - transfer, b + transfer))
    # Pour B -> A
    transfer = min(b, capacity_A - a)
    moves.append((a + transfer, b - transfer))
    return moves
```

```python
from collections import deque
#BFS search
def bfs(start, goal):
    queue = deque([[start]])
    visited = set([start])

    while queue:
        path = queue.popleft()
        a, b = path[-1]
        if a == goal:   # Goal condition: Jug A has 2 liters
            return path
        for state in get_moves((a, b)):
            if state not in visited:
                visited.add(state)
                queue.append(path + [state])
    return None
```

```python
def solve():
    start = (0, 0)
    goal = 2
    solution = bfs(start, goal)
    if solution:
        print("Shortest path to reach 2 liters in Jug A:")
        for step in solution:
            print(step)
        print(len(solution))
    else:
        print("No solution found")
```

```python
solve()
```
