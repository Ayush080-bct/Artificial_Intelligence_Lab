**Problem Statement**
>A farmer has a goat, a wolf and a cabbage on the west side of a river. He wants to transport all three to the east side using a boat that can carry only the farmer and one additional item. The following constraints must always be satisfied:
- The wolf cannot be left alone with the goat.
- The goat cannot be left alone with the cabbage.

```python
#BFS search
state = (0,0,0,0)
goal = (1,1,1,1)
```

```python
def get_moves(state):
    f,g,c,w = state
    return [
        (1-f, g, c, w),        # farmer alone
        (1-f, 1-g, c, w),      # farmer + goat
        (1-f, g, 1-c, w),      # farmer + cabbage
        (1-f, g, c, 1-w)       # farmer + wolf
    ]
```

```python
#constraint defination
def isSafe(state):
    f,g,c,w=state
    if f == 0  :
        if g == 1 and c == 1 :
            return False
        if g == 1 and w == 1:
            return False
    else :
        if g == 0 and c == 0:
            return False
        if g == 0 and w == 0:
            return False
    return True
```

```python
from collections import deque
def bfs_process(start, goal):
    queue = deque([start])
    visited = set()
    stepsToReachGoal = []
    while queue:
        state = queue.popleft()
        if state in visited:
            continue
        if state == goal:
            print("Solution Found Sucessfully")
            break
        visited.add(state)
        stepsToReachGoal.append(state)
        for move in get_moves(state):
            if isSafe(move) and move not in visited:
                queue.append(move)      
    return stepsToReachGoal
```

```python
def describe_move(prev, curr):
    f1,g1,c1,w1 = prev
    f2,g2,c2,w2 = curr
    if g1 != g2: return "Farmer takes Goat"
    if c1 != c2: return "Farmer takes Cabbage"
    if w1 != w2: return "Farmer takes Wolf"
    return "Farmer crosses alone"

def display(start, goal):
    states = bfs_process(start, goal)
    print("Moves:\n")
    for i in range(1, len(states)):
        print(f"Step {i}: {describe_move(states[i-1], states[i])} → {states[i]}")

display(state,goal)
```
